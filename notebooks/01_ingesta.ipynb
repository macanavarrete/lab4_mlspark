## 2. Descarga de Datos

Los datos se descargan desde la API de Socrata correspondiente al dataset **SECOP II - Contratos electrónicos**. Para mejor desarrollo del ejercicio y según el rendimiento del equipo, se limita la descarga a 1.000 registros.

```python
from sodapy import Socrata

client = Socrata("www.datos.gov.co", None)
results = client.get("jbjy-vk9h", query="""
SELECT * 
WHERE
   fecha_de_firma > '2025-11-30T23:59:59'
AND 
    fecha_de_firma < '2026-01-01T00:00:00'
LIMIT   
    100000

    """)
```

```python
len(results)
```

## 3. Carga de datos en Spark

Los datos descargados en formaro JONSON se cargan en un dataframe de spark para su transformación.

```python
# Guardar JSON localmente
json_path = "/opt/spark-data/raw/secop_contratos.json"
os.makedirs(os.path.dirname(json_path), exist_ok=True)

with open(json_path, 'w', encoding='utf-8') as f:
    for record in results:
        f.write(json.dumps(record, ensure_ascii=False) + '\n')

print(f"Datos guardados en: {json_path}")
```

```python
print("Leyendo datos desde JSON...")
df_raw = spark.read.json(json_path)

print(f"Total de registros: {df_raw.count()}")
print(f"Total de columnas: {len(df_raw.columns)}")
```

## 5. Selección de variables clave

Se seleccionaron las columnas mas relevantes para el análisis posterior.

```python
columnas_clave = [
    "referencia_del_contrato",
    "nit_entidad",
    "nombre_entidad",
    "departamento",
    "ciudad",
    "tipo_de_contrato",
    "valor_del_contrato",
    "fecha_de_firma",
    "plazo",
    "plazo_de_ejec_del_contrato",
    "nombre_del_proveedor",
    "estado_contrato"
]
```

```python
columnas_disponibles = [col for col in columnas_clave if col in df_raw.columns]
print(f"\n=== COLUMNAS SELECCIONADAS ({len(columnas_disponibles)}) ===")
for col in columnas_disponibles:
    print(f"- {col}")

# %%

if columnas_disponibles:
    df_clean = df_raw.select(*columnas_disponibles)
else:
    # Si no encontramos las columnas esperadas, usamos todas
    print("ADVERTENCIA: No se encontraron las columnas esperadas. Usando todas las columnas.")
    df_clean = df_raw
```

## 6. Guardar el parquet

La base limpia se almacena en formato parquet, el cual es mas eficiente para el procedimiento. 

```python
output_path = "/opt/spark-data/raw/secop_contratos.parquet"
print(f"\nGuardado en formato parquet")
print(f"Ruta: {output_path}")

df_clean.write \
    .mode("overwrite") \
    .parquet(output_path)

print("Datos guardados exitosamente en formato Parquet")
```

```python
print("\nVerificación")
df_verificacion = spark.read.parquet(output_path)
print(f"Registros en Parquet: {df_verificacion.count():,}")
print(f"Columnas en Parquet: {len(df_verificacion.columns)}")
```
